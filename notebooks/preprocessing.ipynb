# TP2 - Partie 4 : Prétraitement des données

```python


import pandas as pd
import re
import emoji
from collections import Counter
import torch
import numpy as np
```


## Fonction de nettoyage

```python
def clean_tweet(tweet: str) -> str:
    # URL
    tweet = re.sub(r'http\S+', '', tweet)
    
    # Mentions
    tweet = re.sub(r'@\w+', '', tweet)
    
    # Émoticônes (sauf celles standard — on garde les plus courantes pour PLM)
    emoji_pattern = re.compile("["
                               u"\U0001F600-\U0001F64F"  # emoticons
                               u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                               u"\U0001F680-\U0001F6FF"  # transport & map symbols
                               u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                               "]+", flags=re.UNICODE)
    tweet = emoji_pattern.sub(r'', tweet)  # Supprime les emojis non standards
    
    return tweet.strip()
```

## Chargement des données

```python

train_path = "../data/train.csv"
test_path = "../data/test.csv"

df_train = pd.read_csv(train_path)
df_test = pd.read_csv(test_path)



```

## Appliquer le prétraitement

```python
df_train['cleaned_text'] = df_train['tweet'].apply(clean_tweet)
df_test['cleaned_text'] = df_test['tweet'].apply(clean_tweet)
```

## Analyse des artefacts

```python
def analyze_artefact_frequency(df_train: pd.DataFrame, df_test: pd.DataFrame):
    """
    Compte le nombre d'URLs et de mentions @.
    """
    url_count_train = sum(1 for t in df_train['tweet'] if 'http' in t)
    url_count_test = sum(1 for t in df_test['tweet'] if 'http' in t)
    
    mention_count_train = sum(1 for t in df_train['tweet'] if '@' in t)
    mention_count_test = sum(1 for t in df_test['tweet'] if '@' in t)
    
    print(f"URLs : train={url_count_train}, test={url_count_test}")
    print(f"Mentions @ : train={mention_count_train}, test={mention_count_test}")

analyze_artefact_frequency(df_train, df_test)
```

 ## Sauvegarde des données nettoyées

```python
df_train.to_csv("../data/train_cleaned.csv", index=False)
df_test.to_csv("../data/test_cleaned.csv", index=False)

print("✅ Prétraitement terminé — fichiers sauvegardés.")
```
